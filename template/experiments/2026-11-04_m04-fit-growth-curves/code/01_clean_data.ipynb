# Clean the raw growth-curve table

Reads the raw table, keeps the columns this experiment needs under clearer names, checks them, and writes a processed table. The raw file is not changed.

```python
from pathlib import Path

import pandas as pd

RAW = Path("../data/raw/ecoli-growth-37c-1pct.csv")
PROCESSED = Path("../data/processed/ecoli-growth-37c-1pct-clean.csv")

raw = pd.read_csv(RAW)
raw.head()
```

`Average` is the mean of the eight wells, computed in the source workbook. `tcultivo (min)` is the culture time in minutes.

```python
clean = raw[["tcultivo (min)", "T° 600", "Average"]].rename(
    columns={
        "tcultivo (min)": "time_min",
        "T° 600": "temperature_c",
        "Average": "od_mean",
    }
)

assert clean.notna().all().all(), "The table has missing values"
assert clean["time_min"].is_monotonic_increasing, "Time does not increase"
clean.describe()
```

```python
clean.to_csv(PROCESSED, index=False)
print(f"Wrote {PROCESSED}: {len(clean)} rows")
```
