# Repository status

This notebook shows the state of the project's Git repository: its remote, its latest
commits and what has changed since the last commit. Each cell runs a Git command with `!`
and adds what it prints to `../results/repository-status.txt`, so the file keeps the
status with the time the notebook ran. It reads no data.

The exercises are in `M4_INSTRUCTIONS.md` at the top of the project; EX2 runs this
notebook (Run → Run All Cells).

```python
# Start the status file with the time this notebook ran (> writes a new file)
!mkdir -p ../results
!date > ../results/repository-status.txt
```

```python
# The remote: where the project is pushed to and pulled from
# (tee -a shows the output here and adds it to the status file)
!git remote -v | tee -a ../results/repository-status.txt
```

```python
# The latest commits, newest first
!git log --oneline -5 | tee -a ../results/repository-status.txt
```

```python
# What has changed since the last commit (M: modified, ??: untracked)
!git status --short | tee -a ../results/repository-status.txt
```

```python
# The whole status file
!cat ../results/repository-status.txt
```
