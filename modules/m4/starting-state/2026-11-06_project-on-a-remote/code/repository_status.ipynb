# Repository status

This notebook shows the state of the project's Git repository: its remote, its latest
commits, what has changed since the last commit and what Git ignores. It writes this
status to `../results/repository-status.txt`, with the time it ran. It reads no data.

The exercises are in `M4_INSTRUCTIONS.md` at the top of the project; EX2 runs this
notebook (Run → Run All Cells).

```python
# Import the modules: subprocess runs Git, datetime gives the time, pathlib handles paths
import subprocess
from datetime import datetime
from pathlib import Path

PROJECT = Path("../../..")          # work/rrw-workshop, the top of the repository
RESULTS = Path("../results")


def git(*args):
    """Run a Git command in the project and return what it prints."""
    result = subprocess.run(["git", *args], cwd=PROJECT, capture_output=True, text=True)
    return (result.stdout + result.stderr).strip()
```

```python
# The remote: where the project is pushed to and pulled from
remote = git("remote", "-v") or "No remote yet."
print(remote)
```

```python
# The latest commits, newest first
commits = git("log", "--oneline", "-5")
print(commits)
```

```python
# What has changed since the last commit (M: modified, ??: untracked)
changes = git("status", "--short") or "Nothing has changed since the last commit."
print(changes)
```

```python
# What Git ignores: the project's .gitignore
ignored = (PROJECT / ".gitignore").read_text(encoding="utf-8").strip()
print(ignored)
```

```python
# Write the status to a file, with the time this notebook ran
RESULTS.mkdir(exist_ok=True)
now = datetime.now().isoformat(timespec="seconds")
status = f"Status at {now}\n\nRemote:\n{remote}\n\nLatest commits:\n{commits}\n\nChanges:\n{changes}\n"
(RESULTS / "repository-status.txt").write_text(status, encoding="utf-8")
print(f"Wrote {RESULTS / 'repository-status.txt'}")
```
