```python
# Read the growth curve: E. coli at 37 °C with 1 % glucose
import pandas as pd

growth = pd.read_csv("../data/raw/ecoli-growth-37c-1pct.csv")
growth.head()
```

```python
# Culture time in hours, and the optical density as the mean of the eight wells
hours = growth["tcultivo (min)"] / 60
od = growth["Average"]

print(f"{len(growth)} readings over {hours.max():.1f} hours")
print(f"Optical density from {od.min():.3f} to {od.max():.3f}")
```

```python
# Plot the growth curve and save it as a figure of this experiment
from pathlib import Path
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(hours, od)
ax.set_xlabel("Culture time (h)")
ax.set_ylabel("Optical density")
ax.set_title("E. coli at 37 °C, 1 % glucose")

figures = Path("../results/figures")
figures.mkdir(parents=True, exist_ok=True)
fig.savefig(figures / "growth-curve.png", dpi=150)
```
